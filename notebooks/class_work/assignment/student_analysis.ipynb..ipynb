```python
import pandas as pd
import matplotlib.pyplot as plt
import io

csv_data = """Student,Program,Course,Competency,Hours_Studied,Grade
Anna,Informatics,Data Science,Statistics,11,84
Anna,Informatics,Data Science,Programming,12,86
Anna,Informatics,Data Science,Machine Learning,13,90
Anna,Informatics,Data Science,Data Visualization,10,82
Anna,Informatics,Data Science,Big Data,15,93
Benedikt,Education,Education,Pedagogy,14,91
Benedikt,Education,Education,Assessment,12,85
Benedikt,Education,Education,Curriculum Design,13,88
Benedikt,Education,Education,Learning Analytics,14,90
Claudia,Informatics,Data Science,Statistics,9,75
Claudia,Informatics,Data Science,Programming,10,77
Claudia,Informatics,Data Science,Machine Learning,12,82
Claudia,Informatics,Data Science,Data Visualization,8,70
Claudia,Informatics,Data Science,Big Data,11,79
Daniel,Education,Education,Pedagogy,10,72
Daniel,Education,Education,Assessment,9,70
Daniel,Education,Education,Curriculum Design,11,76
Daniel,Education,Education,Learning Analytics,12,78
Elina,Informatics,Data Science,Statistics,14,94
Elina,Informatics,Data Science,Programming,13,91
Elina,Informatics,Data Science,Machine Learning,15,96
Elina,Informatics,Data Science,Data Visualization,12,89
Elina,Informatics,Data Science,Big Data,16,98
Felix,Education,Education,Pedagogy,11,83
Felix,Education,Education,Assessment,10,80
Felix,Education,Education,Curriculum Design,12,84
Felix,Education,Education,Learning Analytics,13,86
Greta,Informatics,Data Science,Statistics,8,73
Greta,Informatics,Data Science,Programming,9,74
Greta,Informatics,Data Science,Machine Learning,11,79
Greta,Informatics,Data Science,Data Visualization,7,68
Greta,Informatics,Data Science,Big Data,10,77
Hanna,Education,Education,Pedagogy,13,88
Hanna,Education,Education,Assessment,12,85
Hanna,Education,Education,Curriculum Design,14,90
Hanna,Education,Education,Learning Analytics,15,92
Ivan,Informatics,Data Science,Statistics,15,97
"""

df = pd.read_csv(io.StringIO(csv_data))
df.head()
```

```python
avg_grade_comp = df.groupby('Competency')['Grade'].mean()
avg_grade_comp.plot(kind='bar', figsize=(8,5), color='skyblue')
plt.title('Keskmine hinne pädevuse lõikes')
plt.ylabel('Grade')
plt.show()
```

The column chart of average grades by competency shows that Machine Learning and Big Data have higher average scores compared to Statistics or Data Visualization. This suggests that students perform better or feel more confident in these competencies.

```python
program_counts = df['Program'].value_counts()

program_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90, figsize=(6,6), colors=['skyblue', 'lightcoral'])
plt.title('Õppijate jaotus programmi järgi')
plt.ylabel('')  # eemaldab y-telje teksti
plt.show()
```

The pie chart shows that most students belong to the Informatics program. This indicates that the dataset has more representation from the Data Science track than the Education track.

```python
plt.figure(figsize=(7,5))
plt.scatter(df['Hours_Studied'], df['Grade'], alpha=0.7, color='mediumseagreen')
plt.title('Seos õppimise tundide ja hinde vahel')
plt.xlabel('Hours Studied')
plt.ylabel('Grade')
plt.show()
```

The scatter plot comparing hours studied versus grades reveals a slight positive correlation. Generally, students who studied more hours tend to achieve higher grades, but there is still some variability among students.

```python
plt.figure(figsize=(7,5))
df.boxplot(column='Grade', by='Program', grid=False, patch_artist=True,
           boxprops=dict(facecolor='lightblue'))
plt.title('Hinnete jaotus programmide lõikes')
plt.suptitle('')
plt.xlabel('Program')
plt.ylabel('Grade')
plt.show()
```

The box plot comparing grades by program indicates that Informatics students have higher median grades and a wider range of scores. Education students have slightly lower grades on average but less variability. This helps identify which program has more consistent performance and which shows more variation

Overall Summary:
The visualizations reveal that students generally perform better in competencies like Machine Learning and Big Data, and those in the Informatics program dominate the dataset. More hours studied tend to correlate with higher grades, although some variability exists. The box plots show that Informatics students have higher and more varied grades, while Education students are slightly lower but more consistent. These insights help understand performance trends across programs, competencies, and study effort.
